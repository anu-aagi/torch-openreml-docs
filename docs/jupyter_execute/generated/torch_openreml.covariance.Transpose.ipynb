```python
import torch
from torch_openreml.covariance import LowerTriangularMatrix, Transpose

op = Transpose(a=LowerTriangularMatrix(3, 2))
free_params = torch.tensor([0.0, 0.5, 1.0, 0.2, -0.3])
grad, grad_names = op.manual_grad(free_params)
grad
```
