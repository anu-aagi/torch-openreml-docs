```python
import torch
from torch_openreml.covariance import EquicorrelationMatrix, HadamardProduct

op = HadamardProduct(
    a=EquicorrelationMatrix(4),
    b=torch.tensor([5.0])
)
free_params = torch.tensor([1.0])
grad, grad_names = op.manual_grad(free_params)
grad
```
